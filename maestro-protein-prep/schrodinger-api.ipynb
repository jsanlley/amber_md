```python
# get path to pdb file
working_dir = '/net/gpfs-amarolab/jsanlleyhernandez/mpro-nat-all/'
test_dir = os.path.join(working_dir,'nsp-4_5')
test_pdb = os.path.join(test_dir,os.listdir(test_dir)[0])
print(test_pdb)
```

```python
# notes on nsp_4_5 (7T70)
# two chains and each peptide substrate has 12 residues
# has solvent fragments (GOL and DMS)
# has crystallographic waters
```

```python
u = mda.Universe(test_pdb) # load universe
                
protein = u.select_atoms("protein") #work with protein residues
for id in set(protein.chainIDs): # calculate the number of residues in a given chain
    chain = protein.select_atoms(f"segid {id}")
    chain_length = len(chain.residues)
    print(f"chain {id}:", chain_length)
    
    if chain_length > 300: # mutate alanine into cysteine (this is for all systems)
        cys = chain.select_atoms("resid 145")
        cys.residues.resnames = "CYS"
        print(cys.residues.resnames)

    if chain_length != 306 and chain_length > 12:
        nterm = "SER" in chain.residues[0].resname
        cterm = "GLU" in chain.residues[-1].resname
        if not cterm:
            # add c termini residue
            
        print(nterm,cterm)
    
solvent = u.select_atoms("not protein") # select solvent molecules (keep xtal waters)
print("solvent:",list(set(solvent.residues.resnames)))


xtal_waters = u.select_atoms("resname HOH and around 5 protein") # filter out crystal waters
print(f"crystal waters {len(xtal_waters.residues)}", '\n')

#combine into one pdb
combined_u = protein + xtal_waters
```
